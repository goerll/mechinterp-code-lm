Goal: test whether SantaCoder internally represents indentation depth and block structure in a way that can be decoded or localized.

Checklist:
- define indentation-sensitive prompt sets
- inspect tokenization of spaces, newlines, and block delimiters
- measure whether indentation depth is linearly decodable
- identify candidate layers and components for deeper circuit analysis
